```python
import matplotlib.pyplot as plt

# Budget data
categories = ["Planned", "Estimated"]
budget = [4000, 2708.22]

# Colors matching the presentation
colors = ["#14243D", "#559DAD"]

# Calculate remaining budget
remaining = budget[0] - budget[1]
percentage = (remaining / budget[0]) * 100

# Create graph
fig, ax = plt.subplots(figsize=(6, 5))

bars = ax.bar(categories, budget, color=colors, width=0.5)

# Display dollar amounts above bars
for bar, value in zip(bars, budget):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value + 100,
        f"${value:,.2f}",
        ha="center",
        fontsize=15,
        fontweight="bold",
        color="#14243D"
    )

# Title
ax.set_title(
    "Planned vs. Estimated Budget",
    fontsize=17,
    fontweight="bold",
    color="#14243D",
    pad=15
)

# Format graph
ax.set_ylim(0, 4800)
ax.set_yticks([])
ax.tick_params(axis="x", labelsize=13, length=0)

# Remove borders
for spine in ax.spines.values():
    spine.set_visible(False)

# Background
fig.patch.set_facecolor("white")
ax.set_facecolor("white")

# Budget evaluation
plt.figtext(
    0.5, 0.06,
    f"Remaining Budget: ${remaining:,.2f}",
    ha="center",
    fontsize=14,
    fontweight="bold",
    color="#14243D"
)

plt.figtext(
    0.5, 0.015,
    f"{percentage:.2f}% Under Budget",
    ha="center",
    fontsize=13,
    color="#14243D"
)

plt.tight_layout(rect=[0, 0.12, 1, 1])

# Save graph
plt.savefig(
    "Team9_Budget_Comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
